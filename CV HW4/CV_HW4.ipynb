### Q1 (10 marks)
Objective: Estimate the disparity map from a stereo image pair using block matching.

You are provided with a stereo image pair (im0.png, im1.png) from the artroom1 scene.

You will use the built-in `cv2.StereoBM_create()` method, which internally uses the Sum of Absolute Differences (SAD):

$$
\text{SAD}(d) = \sum_{i=-k}^{k} \sum_{j=-k}^{k} \left| I_L(x+i, y+j) - I_R(x - d + i, y + j) \right|
$$


---


**Instructions**
1. Load left and right images in grayscale	(1 mark)
2. Correct use of StereoBM_create() with valid parameters	(2 marks)
3. Compute disparity with .compute()	(2 marks)
4. Normalize and visualize the disparity	(3 marks)
5. Clean, readable, and commented code	(2 marks)


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('./HW4 inputs/im0.png', cv2.IMREAD_GRAYSCALE) 
img2 = cv2.imread('./HW4 inputs/im1.png', cv2.IMREAD_GRAYSCALE) 

sift = cv2.SIFT_create() 
# find the keypoints and descriptors with SIFT
kp1, des1 = sift.detectAndCompute(img1,None)
kp2, des2 = sift.detectAndCompute(img2,None)
 
# Create BFMatcher (Brute Force Matcher)
bf = cv2.BFMatcher()
matches = bf.knnMatch(des1,des2, k=2)

pts1 = []
pts2 = [] 
# ratio test as per Lowe's paper
for i,(m,n) in enumerate(matches):
 if m.distance < 0.8*n.distance:
     pts1.append(kp1[m.queryIdx].pt)
     pts2.append(kp2[m.trainIdx].pt)

# Finding Fundamental matrix
pts1_ = np.array(pts1)
pts2_ = np.array(pts2)
F, mask = cv2.findFundamentalMat(pts1_, pts2_, cv2.RANSAC)

# We select only inlier points
pts1_ = pts1_[mask.ravel()==1]
pts2_ = pts2_[mask.ravel()==1]

h,w = img1.shape
# find the two homographies
_, H1, H2 = cv2.stereoRectifyUncalibrated(pts1_, pts2_, F, (w,h))
# Appling the two homographies to the two images to rectify them
rectified_img1 = cv2.warpPerspective(img1, H1, (w,h), flags=cv2.INTER_LINEAR)
rectified_img2 = cv2.warpPerspective(img2, H2, (w,h), flags=cv2.INTER_LINEAR)


# Plots
plt.subplot(121),plt.imshow(rectified_img1, cmap='gray'), plt.axis('off')
plt.subplot(122),plt.imshow(rectified_img2, cmap='gray'), plt.axis('off')
plt.show()


stereo = cv2.StereoBM_create(numDisparities=16, blockSize=21)
disparity = stereo.compute(rectified_img1,rectified_img2)
plt.imshow(disparity,'gray')

```

### Q2 (10 marks)
Compute the depth map from a disparity map using camera calibration parameters.

You are given:

A disparity map (disp0.pfm)

The calibration file (calib.txt) containing the focal length and baseline

Use the formula:

$$
\text{Depth}(x, y) = \frac{f \cdot B}{d(x, y)}
$$


​
**Instructions:**

1. Load disparity map	(1 mark)
2. Parse and extract focal length + baseline	(4 marks)
3. Apply depth formula with safe division	(3 marks)
4. Visualize or display depth map	(2 marks)


```python
import cv2
import numpy as np

# Load the disparity map
import matplotlib.pyplot as plt

disparity = cv2.imread('./HW4 inputs/disp0.pfm', cv2.IMREAD_UNCHANGED)

# Parse the calibration file to extract focal length and baseline
with open('./HW4 inputs/calib.txt', 'r') as f:
    lines = f.readlines()
    for line in lines:
        if 'baseline' in line:
            baseline = float(line.split('=')[1].strip())
    # Extract focal length from the cam0 matrix
    for line in lines:
        if 'cam0' in line:
            cam0_matrix = line.split('=')[1].strip('[]\n').split(';')[0]
            focal_length = float(cam0_matrix.split()[0])
            break

# Compute the depth map using the formula Depth(x, y) = (f * B) / d(x, y)
# Avoid division by zero by replacing zero disparities with a small value
epsilon = 1e-6
disparity[disparity == 0] = epsilon
depth = (focal_length * baseline) / disparity


# Visualize the depth map in grayscale
plt.imshow(depth, cmap='gray')
plt.colorbar(label='Depth')
plt.title('Depth Map')
plt.axis('off')
plt.show()
```
